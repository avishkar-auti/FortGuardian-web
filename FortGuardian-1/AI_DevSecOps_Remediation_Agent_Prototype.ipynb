# AI DevSecOps Remediation Agent Prototype
Built around the existing Prisma X-Ray workflow.

```
architecture='Scheduler -> Image Config -> Prisma Workflow -> Scan Report -> AI Decision -> Artifactory -> Config Update -> Git PR -> Validation -> Teams'; print(architecture)
```

## Configuration

```
image={'IMAGE_NAME':'mta-postfix','IMAGE_TAG':'0.0.4','REGISTRY':'artifactory.sdlc.ctl.gcp.db.com','NAMESPACE':'dkr-public-local/com/db/dbpalace'}
print(image)
```

## Build Image Path

```
path=f"{image['REGISTRY']}/{image['NAMESPACE']}/{image['IMAGE_NAME']}:{image['IMAGE_TAG']}"
print(path)
```

## Trigger Existing Prisma Workflow

```
print('Dispatch prisma-xray-scan-dispatch-wf.yml')
```

## Mock Scan Report

```
scan={'critical':2,'high':5,'medium':4,'low':1}
print(scan)
```

## AI Decision

```
print('Remediation Required' if scan['critical'] or scan['high'] else 'Image Compliant')
```

## Artifactory Recommendation

```
tags=['0.0.4','0.0.5','0.0.6']; rec=tags[-1]; print(rec)
```

## Update Config

```
image['IMAGE_TAG']=rec
print(image)
```

## GitOps

```
for s in ['Create Branch','Commit','Push','Create PR','Attach Evidence']: print('✓',s)
```

## Validation

```
print('Trigger Prisma again and compare results')
```
